```python
# tao tensor tu dong voi gia tri 0 ban dau
# 1 chieu
t_zero = tf.zeros(10)
print(t_zero)

# 2 chieu
t_2_zero = tf.zeros([3,3])
print(t_2_zero)

# 3 chieu
t_3_zero = tf.zeros([3,3,5])
print(t_3_zero)
```

```python
# tao tensor tu dong voi gia tri 1 ban dau
# 1 chieu
t_zero = tf.ones(10)
print(t_zero)

# 2 chieu
t_2_zero = tf.ones([3,3])
print(t_2_zero)

# 3 chieu
t_3_zero = tf.ones([3,3,5])
print(t_3_zero)
```

```python
#tạo tensor tự động với giá trị bất kỳ
# 1 chieu
t_1 = tf.fill(3,10)
print(t_1)

# 2 chieu
t_2 = tf.fill([3,3], 9)
print(t_2)

# 3 chieu
t_3 = tf.fill([3,3,5], 8)
print(t_3)
```

```python
# tạo tensor 1-chiều với hàm linspace trong pham vi
lin_tensor = tf.linspace(5.0, 20.0, 5)
print(lin_tensor)
```

```python
# tạo tensor 1-chiều với hàm range cong them gia tri delta
range_tensor = tf.range(3.0, 7.0, delta=0.5)
print(range_tensor)

range_tensor_1 = tf.range(3.0, delta=0.5)
print(range_tensor_1)
```

```python
# tạo tensor với giá trị ngẫu nhiên
# random_normal
range_tensor = tf.random.normal([5])
print(range_tensor)

range_tensor_1 = tf.random.normal([3,3])
print(range_tensor_1)

range_tensor_2 = tf.random.normal([3,2,2])
print(range_tensor_2)
```

```python
#random.truncated_normal
range_tensor = tf.random.truncated_normal([5])
print(range_tensor)

range_tensor_1 = tf.random.truncated_normal([3,3])
print(range_tensor_1)

range_tensor_2 = tf.random.truncated_normal([3,2,2])
print(range_tensor_2)
#
```

```python
# random.uniform
range_tensor = tf.random.uniform([5], minval=10, maxval=20)
print(range_tensor)

range_tensor_1 = tf.random.uniform([3,2], minval=10, maxval=20)
print(range_tensor_1)

range_tensor_2 = tf.random.uniform([3,2,5], minval=10, maxval=20)
print(range_tensor_2)
```

```python
# ngau nhien voi random.uniform vaf shuffle xao tron gia tri
range_tensor= tf.random.uniform([5],minval=10, maxval= 20)
print(range_tensor)
print(tf.random.shuffle(range_tensor))
```

```python
# chuyen doi kieu du lieu
t_float = tf.constant(3.9, tf.float32)
print(t_float,'---', t_float.dtype)

t_int = tf.cast(t_float,dtype=tf.int32)
print(t_int,'---',t_int.dtype)
```

```python
# cac phep toan
print('Tinh can bac 2')
x = tf.constant(2,dtype=tf.float32)
print(tf.sqrt(x))

print('Cong 2 tensor')
a = tf.constant([[1,2]], tf.int32)
b = tf.constant([[3,4]], tf.int32)
c = tf.add(a,b)
d = a+b
print(c)
print(d)

print('Nhan 2 tensor')
a = tf.constant([[1,2]], tf.int32)
b = tf.constant([[3,4]], tf.int32)
c = tf.multiply(a,b)
d = a*b
print(c)
print(d)
```

```python
# cac ham tinh toan
a = tf.constant([6. , 6. , 6.])
b = tf.constant([4. , 4. , 4.])
c = tf.constant([3. , 3. , 3.])
d = tf.add(a, b)
e = tf.add_n([a, b, c])
f = tf.subtract(a, b)
g = tf.multiply(a, b)
h = tf.divide(a, b)
i = tf.scalar_mul(3, a)
j = tf.math.reciprocal(a)
k = tf.math.mod(a, b)
l = tf.negative(a)

print(d)
print(e)
print(f)
print(g)
print(h)
print(i)
print(j)
print(k)
print(l)
```

```python
# các phép toán làm tròn và so sánh
t = tf.constant([-6.5, -3.5, 3.5, 6.5])
r1 = tf.round(t)
r2 = tf.math.rint(t)
r3 = tf.math.ceil(t)
r4 = tf.math.floor(t)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
# các phép toàn trên vector và ma trận
t1 = tf.constant([4., 3., 2.])
t2 = tf.constant([3., 2., 1.])
t3 = tf.constant([[1, 2, 3],
                  [4, 5, 6]])
t4 = tf.constant([[7, 8],
                  [9, 10],
                  [11, 12]])
dot_x = tf.tensordot(t1, t2, 0)
dot_y = tf.tensordot(t1, t2, 1)
print(t1)
print(t2)
print(t3)
print(t4)
cross = tf.linalg.cross(t1, t2)
mm = tf.matmul(t3, t4)
d = tf.linalg.diag(t3)
trace = tf.linalg.trace(t3)
trans = tf.transpose(t3)
eye = tf.eye(5, 5)
print(dot_x)
print(d)
print(cross)
print(trace)
print(trans)
print(eye)
```
